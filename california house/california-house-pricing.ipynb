```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

Raw_Data = pd.read_csv("/kaggle/input/datasets/sudipgipl/california-housing/train.csv")
print(Raw_Data.isnull().sum())
Home_Data = Raw_Data.fillna(Raw_Data.median(numeric_only=True))
print(Home_Data.dtypes)

Home_Data.columns

y = Home_Data.MedHouseVal

features = ["HouseAge", "AveRooms", "AveBedrms"]
X = Home_Data[features]

DTModel = DecisionTreeRegressor(random_state=1)
train_X, val_X, train_y, val_y = train_test_split(X,y, random_state=1)

DTModel.fit(train_X, train_y)
DTPredictions = DTModel.predict(val_X)

mae = mean_absolute_error(DTPredictions, val_y)
print(mae)

RFModel = RandomForestRegressor(random_state=1)
RFModel.fit(train_X, train_y)
RFPredictions = RFModel.predict(val_X)

mae2 = mean_absolute_error(RFPredictions, val_y)
print(mae2)
print("Error is", mae, "and", mae2)
```
